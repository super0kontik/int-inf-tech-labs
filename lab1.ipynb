```python
import os
import numpy
```

```python
speed = [99,86,87,111,86,103,87,94,78,77,85,86]
x = numpy.mean(speed)

os.system("clear")
print("mean is %3.2f" % x)
```

```python
speed = [99,86,87,111,86,103,87,94,78,77,85,86]
x = numpy.median(speed)
os.system("clear")
print("median is %3.2f" % x)
```

```python
from scipy import stats
```

```python
speed = [99,86,87,111,86,103,87,94,78,77,85,86]
x = stats.mode(speed)
os.system("clear")
print(x)
```
